# P0.4 — Git Basics

## Key Commands
- **git commit** = save changes locally (checkpoint)
- **git push** = send commits to GitHub (remote)
- **git pull** = get latest changes from GitHub to local

## Branch Strategy — RetailPulse
| Branch | Purpose | Who works here |
|--------|---------|----------------|
| dev | Daily development | Always work here |
| uat | Client testing + sign-off | After dev is tested |
| main | Production | Never write directly |

## Rule
Always work in **dev**. Never commit directly to main.
