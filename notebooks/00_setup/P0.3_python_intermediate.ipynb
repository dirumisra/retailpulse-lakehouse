```python
# List Comprehension — build a list using logic in one line

# Regular way — 4 lines

result = []
lst_val = [1, 2, 3, 4, 5]

for x in lst_val:
    result.append(x * 2)
print("Regular way: ", result)
```

```python
# List comprehension — 1 line
lst_val = [1, 2, 3, 4, 5]

result = [x * 2 for x in lst_val]
print("Comprehension:", result)
```

```python
# With condition — only even numbers
evens = [x for x in lst_val if x % 2==0]
print("Evens Only:", evens)
```

```python
# try/except — handle errors without crashing

# Without try/except — this crashes everything
# print(10 / 0)

# With try/except — handled gracefully
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError as e:
    print(f"Error caught: {e}")
    print("Pipeline continues running safely")
```

- *args    = multiple values    → stored as a TUPLE  (immutable)
- **kwargs = multiple key=value → stored as a DICTIONARY

```python
# *args — you don't know how many values coming
def total_price(*args):
    return sum(args)

print(total_price(100, 200, 300))  # 600

# **kwargs — you don't know how many key=value pairs coming
def customer_info(**kwargs):
    for key, values in kwargs.items():
        print(f"{key}: {values}")

customer_info(name="Dhiraj", city="Delhi",profession = "Job", order=5)
```
