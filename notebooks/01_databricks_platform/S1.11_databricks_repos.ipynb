# S1.11 — Databricks Repos
**Date completed:** May 2025  
**Status:** Complete — Used since Day 1

## What Databricks Repos Does
Same as VS Code + Git — but inside Databricks cloud notebook.

| VS Code | Databricks Repos |
|---------|-----------------|
| Write code locally | Write code in cloud notebook |
| git commit | Commit from notebook header |
| git push | Push to GitHub from UI |
| git checkout branch | Switch branch from UI |
| Raise PR on GitHub | Same — PR on GitHub |

## Branch Promotion Workflow
dev → tested → PR raised → uat → approved → PR raised → main

**Rule: Never push directly to main. Always through PR.**

## CI/CD Connection
Databricks Repos is the foundation of our CI/CD pipeline.
Every push to dev → triggers GitHub Actions → promotes to uat → main.
