## Why Lakehouse Was Built — The Evolution

| | Data Warehouse (2000s) | Data Lake (2010s) | Lakehouse (2020s) |
|--|----------------------|------------------|------------------|
| Storage cost | Very expensive | Cheap | Cheap |
| Data types | Structured only | All types | All types |
| ACID transactions | ✅ Yes | ❌ No | ✅ Yes |
| Governance | ✅ Strong | ❌ None | ✅ Strong |
| ML support | ❌ No | ✅ Yes | ✅ Yes |
| SQL speed | ✅ Fast | ❌ Slow | ✅ Fast |
| Schema flexibility | ❌ Rigid | ✅ Flexible | ✅ Flexible |

**Real enterprise pain:**
- Warehouse: Adding one column cost a bank $2M and 6 months
- Data Lake: Netflix data became a "Data Swamp" — nobody trusted it
- Lakehouse: Solved both — cheap storage + ACID + governance + ML
