```python
# We are going to create a bug intentionally
# Then use debugging tools to find it

# Step 1 — Create a simple DataFrame

data = [
    (1, "Dhiru", 2500.0),
    (2, "Rahul", None),
    (3, "Priya", 1800.0),
    (4, "Amit", 3200.0),
    (5, "Sara", None)
]

columns = ["customer_id", "customer_name", "order_value"]

df = spark.createDataFrame(data, columns)
print("DataFrame created successfully")
print(f"Total rows : {df.count()}")
df.show()
```

## **Debugging Tool 1: Data Profiling**

```python
# Debugging Tool 1 — Data Profiling
# Instead of writing multiple lines to check data quality
# One command shows everything

import pyspark.sql.functions as F

print("=== Data Profile ===")
df.describe().show()

print("\n=== Null counts ===")
df.select([
    F.count(F.when(F.col(c).isNull(), c)).alias(c)
    for c in df.columns
]).show()
```

## **Built-in Debugger is the last one.**

```python
# Tool 4 — Built-in Debugger
# To use breakpoints in Databricks:
# 1. Click the line number on the LEFT of any line
# 2. A red dot appears = breakpoint set
# 3. Click the BUG icon (top of notebook) to run in debug mode
# 4. Code pauses at breakpoint — inspect all variables

# Try it — click line number 3 below to set a breakpoint
x = 100
y = 200
z = x + y

print(f"Result: {z}")
```

## Databricks Debugging Tools — 4 Tools

| Tool | How to use | What it shows |
|------|-----------|---------------|
| Error Highlighting | Automatic | Red underlines for syntax errors |
| Data Profiling | df.describe() + null counts | Nulls, distributions, min/max |
| Variable Explorer | Right panel {x} icon | All variables and their values |
| Notebook Debugger | Click line number → bug icon | Pause and inspect at any line |

## Senior Engineer Debugging Approach
1. Read the error message
2. Open Variable Explorer — what state are variables in?
3. Check Data Profiling — is DataFrame empty or has nulls?
4. Set breakpoint at error line — pause and inspect
5. Fix only that cell — rerun just that cell
