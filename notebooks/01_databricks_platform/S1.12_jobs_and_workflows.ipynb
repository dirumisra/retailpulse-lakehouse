**How do you manage pipeline trigger dependencies — Pipeline B runs only after Pipeline A?"**

- In Databricks Jobs — set Task B to depend on Task A.
- Task B only starts when Task A completes successfully.
- If Task A fails — Task B is skipped automatically.
- Configure via the Jobs UI — no code needed.

If any task fails → all downstream tasks are skipped automatically

### 3 Key Concepts

**1 — Task Dependencies**
- Task B only starts when Task A completes successfully
- Configured in Jobs UI — no code needed
- Answer to Q178: set dependency in workflow graph

**2 — Trigger Types**
| Trigger | When it runs |
|---------|-------------|
| Scheduled | Every night at 2am — cron expression |
| File arrival | New file lands in ADLS → pipeline starts |
| API trigger | External system calls Databricks |
| Dependent | After another job completes |
| Manual | Development only — you click Run |

**3 — Retry Logic**
- Task fails → auto retry 3 times
- Still fails → email alert sent
- No human monitoring needed in production

### Task Types Available
Notebook · Python script · SQL · Delta Live Tables · dbt · JAR

### Microsoft Fabric Equivalent
Fabric Pipeline activities with dependency conditions (On success, On failure, On completion) — same concept, different UI
