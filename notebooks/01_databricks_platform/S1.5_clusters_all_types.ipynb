## 1 — All-Purpose Cluster
- Who uses it    : Data engineers during development
- When           : Writing notebooks, testing code, exploring data
- Cost behaviour : Runs until you stop it — auto-terminate after idle
- Enterprise tip : ALWAYS set auto-terminate to 30 mins
- Interview Q    : "Why not use All-purpose in production?" — cost

## 2 — Job Cluster
- Who uses it    : Automated pipelines, scheduled jobs
- When           : Production — runs at scheduled time, dies after
- Cost behaviour : Pay only for runtime — cheapest option
- Enterprise tip : Always use for production pipelines
- Interview Q    : "What cluster does a 2am ETL job use?" — Job cluster

## 3 — SQL Warehouse
- Who uses it    : Data analysts, BI teams, business users
- When           : Running SQL queries, building dashboards, Power BI
- Cost behaviour : Serverless or classic — auto-suspend when idle
- Enterprise tip : Analysts never need a notebook cluster
Interview Q    : "What compute does Databricks SQL Editor use?" — SQL Warehouse

## 4 — Serverless Compute
- Who uses it    : Everyone — 2025 default
- When           : Notebooks, jobs, pipelines — no cluster config needed
- Cost behaviour : Pay per second, starts in seconds
- Enterprise tip : No cluster management = no misconfiguration risk
- Interview Q    : "What is the 2025 default compute in Databricks?" — Serverless
