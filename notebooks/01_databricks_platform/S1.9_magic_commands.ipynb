```python
# Magic Commands — Quick Reference (Python cell)
magic_commands = [
    {"Command": "%md", "Purpose": "Markdown documentation", "Example use": "Headers, tables, notes"},
    {"Command": "%sql", "Purpose": "Run SQL directly", "Example use": "Query Delta tables"},
    {"Command": "%python", "Purpose": "Switch cell to Python", "Example use": "When default is SQL"},
    {"Command": "%sh", "Purpose": "Shell/Linux commands", "Example use": "File operations, installs"},
    {"Command": "%run", "Purpose": "Run another notebook", "Example use": "Import shared functions"},
    {"Command": "%pip", "Purpose": "Install packages", "Example use": "%pip install faker"},
]

import pandas as pd
display(pd.DataFrame(magic_commands))

# Key Difference
key_diff = [
    {"Command": "%run", "Description": "imports notebook as module, variables are shared"},
    {"Command": "dbutils.notebook.run()", "Description": "runs as separate job, returns result string"},
]
display(pd.DataFrame(key_diff))

# Faker library installed
# Will be used later to generate test data for RetailPulse pipeline testing
```
