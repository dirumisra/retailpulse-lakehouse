## S1.6 — Cluster Configuration
**Date completed:** May 2025  
**Interview covered:** Q218 — How to right-size a cluster

## The 3 Rules Senior Engineers Remember

1. **Match size to data** — 500MB needs 2 workers, not 10
2. **Job cluster for production** — starts, runs, terminates, cheapest
3. **Spot = development only** — On-Demand for critical 2am pipelines

## Right-Sizing Guide
| Data Size | Worker Nodes |
|-----------|-------------|
| < 1 GB | 2–4 workers |
| 1–100 GB | 4–8 workers |
| 100 GB+ | 8–32+ workers |
| Streaming | Autoscaling always |

## 4 Key Decisions
- **Runtime:** Standard (DE work) vs ML Runtime (ML work)
- **Node size:** Driver = 2x RAM of workers
- **Autoscaling:** ON for streaming, OFF for fixed batch jobs
- **Instance:** On-Demand for production, Spot for development
