```python
# Module 3 — dbutils.widgets — Notebook parameters

dbutils.widgets.text("environment", "dev", "Select Environment")
dbutils.widgets.text("date", "2025-05-01", "Processing Date")

# Read the widget values
env = dbutils.widgets.get("environment")
date = dbutils.widgets.get("date")

print(f"Running pipeline for environment: {env}")
print(f"Processing date: {date}")
print(f"This notebook now works in dev, uat, and prod without changing code")
```
