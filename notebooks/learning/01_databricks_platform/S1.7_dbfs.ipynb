# S1.7 — DBFS (Databricks File System)
**Date completed:** May 2025  
**Status:** Complete — Legacy awareness topic

## What DBFS Does
1. Shortens long cloud storage paths
2. Unifies different storage (Azure, AWS) under one path style
3. Abstracts the cloud provider underneath

## Old vs New
| | Old (DBFS) | New (Unity Catalog Volumes) |
|--|-----------|---------------------------|
| Path | dbfs:/mnt/retailpulse/ | /Volumes/retailpulse/ |
| Governance | None | Full RBAC + lineage |
| Status | Legacy — being retired | Modern standard |

## Why Learn It
Legacy code uses dbfs:/mnt/ everywhere.
You must read it, understand it, migrate it.
