## What is Tungsten?

Catalyst decides WHAT to execute.  
Tungsten decides HOW to execute it at machine level.

## The Problem Tungsten Solves
Java object overhead wastes memory:
- Integer 42 in Java = 16 bytes (12 bytes wasted overhead)
- 100M integers = 1.6 GB with Java vs 400 MB with Tungsten
- More RAM needed = bigger machines = higher cost

## 3 Tungsten Solutions

**1 — Off-heap Binary Format**
- Raw binary bytes instead of Java objects
- 75% less memory usage
- No garbage collection overhead

**2 — Code Generation**
- Compiles query to bytecode before executing
- No interpretation at runtime
- Like memorising a recipe vs reading while cooking

**3 — Cache-aware Computation**
- Arranges data to fit in CPU L1/L2 cache
- CPU speed processing instead of RAM speed
- Dramatically faster aggregations
