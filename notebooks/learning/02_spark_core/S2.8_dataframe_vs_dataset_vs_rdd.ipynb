```python
# ============================================================
# Cell 2 — DataFrame vs Dataset vs RDD
# Goal: See each approach on the same data
# Simulating all 3 since SparkContext not available in serverless
# ============================================================

import pyspark.sql.functions as F
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, LongType

print("=== DataFrame vs Dataset vs RDD ===")
print()

# --- Situation 1: Unstructured log data → RDD approach ---
print("--- Situation 1: Unstructured logs → RDD approach ---")
print("Raw log line: '192.168.1.1 GET /api/orders 200 1234'")
print("No fixed schema → RDD is the right tool")
print("# rdd = sc.textFile('logs.txt')")
print("# parsed = rdd.map(lambda line: line.split(' '))")
print("# errors = parsed.filter(lambda x: x[3] == '500')")
print("SparkContext not available in serverless — concept only")
print()

# --- Situation 2: Structured data → DataFrame approach ---
print("--- Situation 2: Structured Olist data → DataFrame ---")

# Define schema explicitly — best practice in production
schema = StructType([
    StructField("order_id", LongType(), False),
    StructField("customer_id", LongType(), False),
    StructField("city", StringType(), True),
    StructField("order_value", DoubleType(), True)
])

# Create sample data
data = [
    (1001, 201, "Delhi", 2500.0),
    (1002, 202, "Mumbai", 1800.0),
    (1003, 203, "Pune", 3200.0),
    (1004, 204, "Delhi", 950.0),
    (1005, 205, "Chennai", 4100.0)
]

df = spark.createDataFrame(data, schema)

print("Schema enforced — wrong types caught at creation time:")
df.printSchema()
df.show()

# --- Situation 3: Type safety → Dataset (Scala only) ---
print("--- Situation 3: Type safety → Dataset ---")
print("Dataset[Transaction] available in Scala/Java only")
print("In Python: DataFrame IS our Dataset")
print("Wrong type in Python DataFrame → caught at runtime")
print("Wrong type in Scala Dataset → caught at COMPILE time")
print()
print("Enterprise rule: Python teams → DataFrame always")
print("                 Scala teams → Dataset for domain objects")
```

## Key Takeaways — S2.8

## Decision Framework
| Situation | Choose | Why |
|-----------|--------|-----|
| Unstructured data (logs, binary) | RDD | No schema possible |
| Structured data in Python | DataFrame | Catalyst + simple syntax |
| Structured data in Scala/Java + type safety | Dataset | Compile-time checking |

## Enterprise Reality (2026)
- 95% of work → DataFrame
- 4% of work → RDD (legacy or unstructured only)
- 1% of work → Dataset (Scala teams only)

## Schema Enforcement
Defining schema explicitly = best practice
- Wrong types caught at creation time
- nullable = false → column cannot be null
- nullable = true → null values allowed
- First step toward data quality (covered in S11)

## Python Rule
In Python — DataFrame IS your Dataset.
There is no Dataset API in PySpark.
Use DataFrame for everything structured.
